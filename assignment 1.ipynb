```python
def numpy_spiral_order(matrix):
    result = []
    top, bottom = 0, matrix.shape[0] - 1
    left, right = 0, matrix.shape[1] - 1

    while top <= bottom and left <= right:
        # Left to right
        for i in range(left, right + 1):
            result.append(matrix[top][i])
        top += 1

        # Top to bottom
        for i in range(top, bottom + 1):
            result.append(matrix[i][right])
        right -= 1

        if top <= bottom:
            # Right to left
            for i in range(right, left - 1, -1):
                result.append(matrix[bottom][i])
            bottom -= 1

        if left <= right:
            # Bottom to top
            for i in range(bottom, top - 1, -1):
                result.append(matrix[i][left])
            left += 1

    return result


# Test the function
spiral = numpy_spiral_order(arr)
print("Spiral Order:", spiral)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("videogamesales.csv")
df.head()
```

a)

```python
df['Global_Sales'] = (
    df['NA_Sales'] +
    df['EU_Sales'] +
    df['JP_Sales'] +
    df['Other_Sales']
)

df_sorted = df.sort_values(by='Global_Sales', ascending=False)
print(df_sorted[['Name', 'Global_Sales']].head())
```

b)

```python
genre_sales = df.groupby('Genre')['Global_Sales'].sum()

plt.figure(figsize=(10, 6))
genre_sales.plot(kind='bar')
plt.title("Total Global Sales by Genre")
plt.xlabel("Genre")
plt.ylabel("Total Sales (Millions)")
plt.xticks(rotation=45)
plt.show()
```

c)

```python
gta_games = df[df['Name'].str.contains('Grand Theft Auto', case=False, na=False)]

gta_df = gta_games[['Name', 'Platform', 'Year', 'EU_Sales', 'JP_Sales']]
gta_df['EU_JP_Sales'] = gta_df['EU_Sales'] + gta_df['JP_Sales']

gta_df = gta_df[['Name', 'Platform', 'Year', 'EU_JP_Sales']]
gta_df
```

d)

```python
gta_sales = {
    'North America': gta_games['NA_Sales'].sum(),
    'Europe': gta_games['EU_Sales'].sum(),
    'Japan': gta_games['JP_Sales'].sum(),
    'Other': gta_games['Other_Sales'].sum()
}

plt.figure(figsize=(7, 7))
plt.pie(gta_sales.values(), labels=gta_sales.keys(), autopct='%1.1f%%')
plt.title("Grand Theft Auto Sales by Region")
plt.show()
```
